# Seleksi Fitur

- Anda dapat menggunakan model pipeline untuk mempermudah perkejaan Anda.

- Berdasarkan hasil analisa Anda, berapa jumlah fitur terbaik yang dapat digunakan? Apa saja fitur tersebut?

### Import library

```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.feature_selection import chi2, SelectKBest, RFE
from sklearn.metrics import accuracy_score, classification_report
```

```python
# Pisahkan Diagnosis 
y = df["diagnosis"].astype(int)
X = df.drop(columns=["diagnosis"])
```

# Setup Pipeline Preprocessing & Train- Test Split

```python
print(list(X.columns))
```

```python
from sklearn.feature_selection import f_classif

# Karena seluruh kolom X adalah numerik:
num_cols = list(X.columns)

# Pipeline preprocessing numerik (Scaling)
preprocess = ColumnTransformer([
    ("num", StandardScaler(), num_cols)
])

# Split dataset (80% train, 20% test, stratify berdasarkan y)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, stratify=y, test_size=0.2, random_state=42
)
```

```python
best_k = 0
best_acc = 0
results = []

for k in range(1, len(num_cols) + 1):
    pipe = Pipeline([
        ("prep", preprocess),
        ("sel", SelectKBest(score_func=f_classif, k=k)),
        ("clf", LogisticRegression(max_iter=1000, random_state=42))
    ])
    
    pipe.fit(X_train, y_train)
    pred = pipe.predict(X_test)
    acc = accuracy_score(y_test, pred)
    results.append((k, acc))
    
    if acc > best_acc:
        best_acc = acc
        best_k = k

print(f"Jumlah Fitur Terbaik (k): {best_k} dengan Akurasi: {best_acc:.4f}")
```

```python
# Pipeline dengan k terbaik
final_pipe = Pipeline([
    ("prep", preprocess),
    ("sel", SelectKBest(score_func=f_classif, k=best_k)),
    ("clf", LogisticRegression(max_iter=1000, random_state=42))
])

final_pipe.fit(X_train, y_train)
pred_final = final_pipe.predict(X_test)

print("=== Classification Report ===")
print(classification_report(y_test, pred_final))
```

```python
# Ambil nama fitur dari preprocessing
feat_names = final_pipe.named_steps["prep"].get_feature_names_out()

# Ambil mask dan skor dari SelectKBest
sel = final_pipe.named_steps["sel"]
mask = sel.get_support()
selected_names = feat_names[mask]
selected_scores = sel.scores_[mask]

# Urutkan berdasarkan skor tertinggi
top_features = sorted(zip(selected_names, selected_scores), key=lambda t: t[1], reverse=True)

print(f"\n--- {best_k} Fitur Terpilih (Diurutkan dari Skor Tertinggi) ---")
for idx, (name, score) in enumerate(top_features, 1):
    clean_name = name.replace("num__", "")
    print(f"{idx}. {clean_name:<25} | Skor: {score:.4f}")
```

```python
# from sklearn.feature_selection import RFE

# RFE dengan Logistic Regression
rfe_pipe = Pipeline([
    ("prep", preprocess),
    ("sel", RFE(estimator=LogisticRegression(max_iter=1000, random_state=42), n_features_to_select=best_k)),
    ("clf", LogisticRegression(max_iter=1000, random_state=42))
])

rfe_pipe.fit(X_train, y_train)
pred_rfe = rfe_pipe.predict(X_test)
print(f"Akurasi dengan RFE (k={best_k}):", accuracy_score(y_test, pred_rfe))
```
